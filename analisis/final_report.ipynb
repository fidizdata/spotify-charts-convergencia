# 1. Introducción: ¿Globalización musical?

Vivimos en la era de la disponibilidad total. Por primera vez en la historia, un usuario en Tokio y otro en Buenos Aires tienen acceso al mismo catálogo de millones de canciones en la misma plataforma. Bajo esta premisa, la lógica del mercado sugeriría una convergencia inevitable: una completa globalización musical, donde las listas de éxitos serían idénticas en todo el globo.

El acceso a la información ha instalado la creencia de que vivimos en un mundo culturalmente uniforme. La narrativa predominante sugiere que, al democratizar el acceso a la música, las plataformas de *streaming* actuarían como un disolvente de las identidades locales, forzando una convergencia hacia un gusto global estandarizado. Posiblemente con un gran peso de los centros de producción hegemónicos, principalmente la industria anglosajona.

Bajo esta premisa, el *Top 200* de un país como Argentina debería ser cada vez más indistinguible del de Japón o Noruega. Sin embargo, ¿es esta la realidad que reflejan los datos? ¿Estamos realmente asistiendo al nacimiento de una globalización musical homogénea?

### 1.1. Objetivo del estudio
Este informe se propone evaluar el estado de la globalización cultural a través del análisis de consumo de Spotify en el período 2017–2021. Nuestra meta no es solo describir qué canciones son populares, sino diagnosticar si el mercado global tiende a una uniformidad en su consumo o si, por el contrario, los paises se diferencian entre sí y se agrupan en torno a afinidades regionales.
Para ello, el informe responde tres preguntas:

1. ¿Existe una convergencia real? ¿Están los países escuchando música cada vez más parecida?
2. ¿Cómo se organiza el mundo musical? ¿Es la geografía el factor determinante, o existen "bloques de afinidad" que ignoran las fronteras tradicionales? ¿Y esa organización cambia durante el período?

### 1.2. Marco metodológico
Para responder a estas interrogantes, abandonamos la mera descripción de las listas de éxitos y adoptamos un enfoque sistémico. La complejidad del fenómeno requiere herramientas de **análisis de redes y análisis de convergencia**:

1. **Naturaleza del dato:** Los datos provienen del histórico de *Spotify Charts* (2017–2021). Se conservaron solo los rankings Top 200 (se descartó el Viral 50) y, para cada mercado y año, se sumaron los streams que cada canción registró en esos rankings y se seleccionaron las 200 con más streams. Cada canción se identifica por su título y su artista. Se mantuvieron los mercados con datos en los cinco años y se excluyeron "Global" y tres mercados con datos incompletos (Bulgaria, Luxemburgo y Nicaragua). El panel resultante comprende 53 mercados con 200 canciones por mercado y año (53.000 registros).

2. **Representatividad del streaming:** Al centrarse en los 200 temas con mayor volumen de reproducción, este dataset no representa la totalidad del consumo cultural de una nación, sino su **repertorio *mainstream***. Esto es metodológicamente relevante, ya que el *Top 200* es el campo de batalla donde convergen las **estrategias de marketing global**, las **preferencias masivas** y los **algoritmos de recomendación**. Aunque no abarca el consumo de nicho o fuera de plataformas, este dataset es el **estándar de facto de la industria musical actual**. Por lo tanto, cualquier divergencia o convergencia observada aquí tiene implicancias directas sobre cómo se está moldeando el gusto público y la economía de la música a escala global.

Sobre esta base, aplicamos las siguientes técnicas:

1. **Cuantificación mediante Distancia Coseno:** Definimos el perfil musical de cada país como un **vector de alta dimensionalidad** en el espacio de canciones disponibles. Cada dimensión representa una canción y su magnitud es la **cuota de mercado (*stream share*)** que esta ocupa en el consumo nacional. La **Distancia de Coseno** mide el ángulo entre estos vectores: si el ángulo es pequeño (distancia cercana a 0), los países comparten no solo el catálogo, sino las mismas prioridades e intensidades de escucha.
2. **Marco de Convergencia ($\sigma$-convergencia):** Inspirándonos en los modelos de convergencia económica de **Barro y Sala-i-Martin**, adaptamos el concepto de **$\sigma$-convergencia** al dominio cultural. Evaluamos si la **dispersión o distancia bilateral media** entre los perfiles musicales de las naciones experimenta una contracción sostenida a lo largo del tiempo.
3. **Modelado Topológico (MST):** Modelamos el mercado como un grafo ponderado de afinidades. Mediante el **Árbol de Expansión Mínima (MST)**, aislamos el **esqueleto estructural del sistema** (las conexiones estrictamente necesarias de mínima distancia), eliminando aristas redundantes para identificar el vecino más cercano de cada país y las conexiones entre regiones.
4. **Detección de comunidades:** Implementamos el algoritmo de **Greedy Modularity** sobre el grafo enriquecido con aristas del **percentil 90 de similitud**, lo que permite identificar si el sistema se particiona de manera natural en **bloques culturales o continentes musicales**.

# 2. La Divergencia del Consumo Global

Para evaluar el grado de integración musical, hemos definido la distancia entre dos países mediante la **Distancia de Coseno** ($1 - \text{Similitud de Coseno}$), aplicada sobre la cuota de mercado (*stream share*) de cada canción.

### 2.1. El fenómeno de $\sigma$-Convergencia
Para diagnosticar si el sistema tiende a la homogeneidad, analizamos la evolución temporal de la **Distancia Coseno Media ($\bar{d}_t$)** de todas las duplas de países, un indicador adaptado de la $\sigma$-convergencia. Si el *streaming* fuera una fuerza democratizadora, observaríamos una caída sostenida en estas distancias.



```python
from src.network_analysis import calculate_cosine_similarity_matrix, build_convergence_table, calculate_sigma_convergence

# 1. Construir tabla de distancias usando Coseno
df_dist_coseno = build_convergence_table(df_top200_annual, calculate_cosine_similarity_matrix)

# 2. Calcular estadísticas de convergencia
sigma_df = calculate_sigma_convergence(df_dist_coseno)

# 3. Graficar
plt.figure(figsize=(10, 5), dpi=120)
sns.lineplot(
    data=sigma_df, 
    x="year", 
    y="mean_distance", 
    marker="o", 
    color="#d95f02", 
    linewidth=2.5
)

plt.title("Evolución de la Distancia Musical Global (Coseno, 2017–2021)", fontsize=13, fontweight="bold", pad=15)
plt.xlabel("Año", fontsize=11)
plt.ylabel("Distancia Promedio (1 - Similitud)", fontsize=11)
plt.xticks([2017, 2018, 2019, 2020, 2021])
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Ocultar el índice y formatear con 2 decimales las columnas deseadas
sigma_df.style.hide(axis="index").format(
    {
        "mean_distance": "{:.2f}",
        "std_distance": "{:.2f}",
    }
)
```



La distancia coseno media pasa de 0,48 en 2017 a 0,72 en 2021, un aumento de aproximadamente 50 %. La serie es: 0,48 (2017), 0,66 (2018), 0,72 (2019), 0,70 (2020) y 0,72 (2021). En el marco de la $\sigma$-convergencia, esto no respalda la homogeneización: la distancia no se contrajo, aumentó.


### 2.2. Hallazgos: **Diversidad estructural** y **evolución temporal de la distancia**

Los datos muestran que el consumo musical global analizado no constituye un **sistema homogéneo**. La **distancia media** está lejos de acercarse a 0 y la **dispersión** entre pares de países se mantiene alta de manera constante (con una desviación estándar que oscila entre 0,24 y 0,26 en todos los años), lo que confirma que existen **diferencias estructurales** profundas y persistentes entre los distintos mercados.

Al observar la trayectoria de la distancia media a lo largo de la serie temporal, los datos dividen el período en dos etapas diferenciadas por su tendencia:

1. **Fase de divergencia (2017–2019):** Se registra un **incremento sostenido de la distancia media** entre los mercados musicales, un proceso en el cual ocurre el **97% del aumento total acumulado** en todo el período analizado. Matemáticamente, esto indica que los charts nacionales se volvieron más distantes y diferenciados entre sí.
2. **Fase de estabilización (2020–2021):** El indicador interrumpe su crecimiento y alcanza una **meseta**, estabilizando los niveles de distancia promedio en un rango de entre **0,70 y 0,72**. Esto refleja que el grado de diferenciación entre los mercados alcanzó un **punto de equilibrio** estacional o estructural dentro de la ventana temporal estudiada.



Para comprobar que el resultado no depende de cómo se mide la distancia, repetimos el cálculo con otras dos definiciones: coseno con `log1p`, que atenúa el peso de las canciones más escuchadas, y Jaccard, que solo considera qué canciones aparecen en el Top 200 de cada país, sin volúmenes.

Las tres definiciones aumentan con el mismo patrón: de 0,48 a 0,72 con coseno sobre *stream share*, de 0,55 a 0,77 con coseno `log1p` y de 0,70 a 0,86 con Jaccard.


Los valores iniciales confirman la ausencia de un consumo homogéneo a nivel global, y todas las mediciones replican de forma consistente las **dos fases estructurales** identificadas (el proceso de divergencia continua seguido por la estabilización o meseta hacia el tramo final del período). Esto demuestra que las conclusiones sobre la fragmentación y el equilibrio de los mercados son independientes de la parametrización matemática elegida

### 2.3. Interpretación: ¿Por qué diverge el sistema?

Si las plataformas de streaming ofrecen un acceso global a catálogos masivos en todo el planeta, la tendencia esperable bajo un supuesto de uniformidad sería la convergencia. Sin embargo, los datos muestran el proceso opuesto: **divergencia**. 

Esto indica que los mercados no operan como un sistema integrado que se homogeneiza con el tiempo. Por el contrario, la ampliación de la distancia media sugiere que las dinámicas de consumo —mediadas por catálogos locales y sistemas de recomendación que operan a nivel regional— amplifican las diferencias estructurales entre los charts nacionales en lugar de reducirlas.

# 3. Estructura del las comunidades musicales: Análisis de Redes y Comunidades

Habiendo descrito que el sistema musical global diverge, el siguiente paso metodológico consiste en ver cómo se organiza esta diversidad. Si el mundo no es una unidad homogénea, ¿cuáles son sus fracturas, sus bloques de afinidad y sus conexiones? 1. **El espacio relacional puro (El MST):** Extraemos el Árbol de Expansión Mínima para simplificar la red global y aislar las conexiones bilaterales más fuertes (la "columna vertebral" del consumo), basándonos exclusivamente en la distancia de coseno y prescindiendo de la geografía.
2. **La estructura de comunidades (La Red):** Aplicamos algoritmos de detección de comunidades sobre el grafo para identificar agrupamientos latentes (clusters de países con patrones de escucha altamente homogéneos) a partir de la topología intrínseca. Proyectamos las comunidades identificadas sobre el mapa geográfico para evaluar en qué medida las fronteras culturales coinciden con los límites continentales o responden a dinámicas transnacionales.

> **Nota metodológica sobre el año de referencia (2021):**

El MST nos permite observar la "columna vertebral" del consumo musical.
En la figura, el tamaño de cada nodo crece con el volumen de streams del Top 200 del mercado, lo que permite ver no solo qué países se parecen, sino también cuáles pesan más en el panel.

# 4. Análisis de la Estabilidad Estructural (2017-2021)

Al examinar la composición de los grupos a lo largo del período, se observa que la organización de los mercados musicales no ha sido inmutable, aunque el cambio no es parejo: dos de los tres grupos mantienen casi sin cambios su composición (Asia suma a Tailandia en 2018) y la variación está en el resto. La composición cambia en cada paso entre 2017 y 2020, y la de 2020 es idéntica a la de 2021.


```python
import plotly.express as px

def generar_y_graficar_todos_los_anios(df_top200_annual, percentile_cutoff=90):
    """
    Detecta clústeres y genera el mapa coroplético para cada año, 
    centrado en África, sin zoom, sin barra de herramientas, con la latitud ajustada en 25 
    y una paleta de colores fija por clúster.
    """
    # Obtener la lista ordenada de todos los años disponibles
    anios = sorted(df_top200_annual['year'].unique())
    
    print(f"Se encontraron {len(anios)} años para procesar: {anios}\n")
    
    # Mapeo de colores fijo:
    # Cluster 1: azul, Cluster 2: verde, Cluster 3: violeta, Cluster 4: rojo, Cluster 5: naranja
    colores_clusters = {
        "Cluster_1": "#1f77b4",  # Azul
        "Cluster_2": "#2ca02c",  # Verde
        "Cluster_3": "#9467bd",  # Violeta
        "Cluster_4": "#d62728",  # Rojo
        "Cluster_5": "#ff7f0e"   # Naranja
    }
    
    for anio in anios:
        print(f"--- Procesando año {anio} ---")
        
        # 1. Detectar clústeres para el año actual
        df_clusters_anio, _ = detect_musical_clusters(
            df_top200=df_top200_annual, 
            year=anio, 
            percentile_cutoff=percentile_cutoff,
            include_mst=True
        )
        
        # Verificar que el DataFrame no esté vacío
        if df_clusters_anio.empty:
            print(f"No hay datos suficientes para el año {anio}.\n")
            continue
            
        # 2. Preparar los datos para Plotly
        n_clusters = df_clusters_anio["cluster_id"].nunique()
        percentil = df_clusters_anio["percentile_cutoff"].iloc[0]
        
        df_plot = df_clusters_anio.copy()
        df_plot = df_plot.rename(columns={'region': 'pais', 'cluster_size': 'integrantes_cluster'})
        df_plot['cluster_id'] = df_plot['cluster_id'].astype(str)
        
        # Asegurar que tengan el prefijo exacto para coincidir con el diccionario de colores (ej. "Cluster_1")
        if not df_plot['cluster_id'].iloc[0].startswith("Cluster_"):
            df_plot['cluster_id'] = "Cluster_" + df_plot['cluster_id']
            
        # 3. Crear la figura con los colores fijos asignados
        fig = px.choropleth(
            df_plot,
            locations="pais",
            locationmode="country names",
            color="cluster_id",
            hover_name="pais",
            hover_data={"integrantes_cluster": True, "cluster_id": True, "year": True},
            title=f"Clústeres de Consumo Musical — Año {anio} (Percentil {percentil}%) — Total: {n_clusters} Clústeres",
            color_discrete_map=colores_clusters # <--- Asigna los colores fijos por ID de clúster
        )
        
        # Configuración con África en el centro, el desplazamiento vertical ajustado a 25 y sin cortes en el norte
        fig.update_layout(
            geo=dict(
                showframe=False, 
                showcoastlines=True, 
                projection_type="natural earth",
                center=dict(lat=25, lon=20), # 
                projection_rotation=dict(lon=20)
            ),
            dragmode=False,
            legend_title_text="ID de Clúster",
            margin={"r":0, "t":50, "l":0, "b":0}
        )
        
        # Mostrar el gráfico sin zoom por scroll ni barra de herramientas
        fig.show(config={
            'scrollZoom': False,
            'displayModeBar': False,
            'doubleClick': False
        })

# Ejecutar el proceso completo para todos los años:
generar_y_graficar_todos_los_anios(df_top200_annual, percentile_cutoff=90)
```

### 5.1. Síntesis de hallazgos
El análisis del consumo mainstream de Spotify en 53 mercados (2017–2021) arroja cinco resultados empíricos:

1. **Divergencia sostenida y estabilización.** La distancia coseno media entre países aumentó ~50 % (0,48 → 0,72) entre 2017 y 2021. El 97 % del incremento ocurrió en 2017–2019; 2020–2021 muestra una meseta (0,70–0,72). El patrón se replica con métricas *cosine log1p* y *Jaccard*, confirmando que no es un artefacto de la definición de distancia.

2. **Contracción del repertorio compartido (canciones globales).** Las canciones presentes en ≥20 mercados cayeron de 138 (2017) a 74 (2021); su participación en los streams pasó de 56 % a 28 %. La concentración local (top 10 por país) permaneció estable en 14 %–15 %. La fragmentación no proviene de listas más cerradas, sino de **menor superposición internacional** de los éxitos.

3. **Estructura modular predominantemente regional.** El Árbol de Expansión Mínima (MST) —el esqueleto de 52 aristas que minimiza la distancia global— coloca **43 de 52 conexiones (83 %) dentro de la misma macrorregión** (América Latina, Europa, Norteamérica, Oceanía, Asia). Bajo aleatoriedad se esperan ~17 (test de permutación *p* < 0,001). La geografía física es el principal organizador de la similitud musical a gran escala.

4. **Tres comunidades con distinta cohesión y estabilidad (2021).**  
   * **Iberoamérica + España (16 países):** única comunidad de alta cohesión (distancia intra-grupo 0,22 vs 0,72 global). Composición idéntica los 5 años, membresia temporal constante.  
   * **Asia (8 países):** comunidad geográficamente pura, pero cohesión baja (distancia intra-grupo 0,68). Composición estable desde 2018, membresia remporal relativamente constante
   * **Anglo-Europa + Brasil + Turquía (29 países):** comunidad extensa, heterogénea (std 0,20) y la única inestable al variar parámetros o método, es decir que la membresia de sus paises no es constante.


6. **Puentes culturales específicos en el MST.** Las 9 aristas inter-regionales del MST revelan tres patrones no geográficos:  
   * **Portugal** es nodo bisagra lusófono (une a Brasil y Costa Rica con Europa), pero **no conecta con España** (España va a Perú).  
   * **Núcleo anglófono** de distancias ultracortas: Australia–Canadá (0,15), Australia–Irlanda (0,19), EE. UU.–Canadá (0,15).



### 5.2. Limitaciones
Estos hallazgos están acotados por el diseño del estudio:

* **Ventana temporal (5 años).** No permite distinguir si la meseta 2020–2021 es un equilibrio estructural o una pausa transitoria.
* **Panel de 53 mercados.** Resultados válidos solo para los mercados de Spotify con serie completa; no generalizables al conjunto de naciones.
* **Definición de comunidad.** La partición depende del grafo (MST + percentil 90) y del algoritmo (Greedy Modularity). Con otros percentiles el número de clusters 2021 varía de 2 a 5; el análisis prioriza composición sobre conteo.
* **Solo consumo (lado de la demanda).** No se observan variables de oferta (catálogo local, inversión A&R, campañas), ni metadatos de artista (nacionalidad, género, idioma de la letra).
* **Granularidad del dato.** Top 200 anual agregado; remixes/versiones con distinta ortografía cuentan como canciones distintas; no hay datos demográficos de oyentes.
